### №1 Two sum

```python
def twoSum(nums, target):
    left = 0
    right = len(nums) - 1
    while left<right:
        s = nums[left] + nums[right]
        if s==target:
            return left, right
        elif s < target:
            left+=1
        else:
            right-=1
    return left, right
```

```python
nums = [3, 8, 9, 11, 16, 18, 19, 21]
print(twoSum(nums, 25))
```

### №2 Развернуть массив

```python
def reverseArray(arr):
    left = 0
    right = len(nums) - 1
    while left<right:
        arr[left], arr[right] = arr[right], arr[left]
        left+=1
        right-=1
    return arr
```

```python
nums = [3, 8, 9, 11, 16, 18, 19, 21, 25]
print(reverseArray(nums))
```

### №3 Развернуть часть массива

```python
def reversePartArray(arr, left, right):
    while left<right:
        arr[left], arr[right] = arr[right], arr[left]
        left+=1
        right-=1
   
def solution(arr, k):
    n = len(arr)
    k = k%n
    reversePartArray(arr, 0, n-1)
    reversePartArray(arr, 0, k-1)
    reversePartArray(arr, k, n-1)
    return arr
```

```python
nums = [3, 8, 9, 11, 16, 18, 19, 21, 25]
print(solution(nums, 5))
```

### №4 Слияние двух отсортированных массивов

```python
def merge_sorted_arr(arr1, arr2):
    merged_arr = []
    i = 0
    j = 0
    while i<len(arr1) and j<len(arr2):
        if arr1[i]<arr2[j]:
            merged_arr.append(arr1[i])
            i+=1
        else:
            merged_arr.append(arr2[j])
            j+=1
    merged_arr.extend(arr1[i:])
    merged_arr.extend(arr2[j:])
    return merged_arr
```

```python
arr1 = [3, 7, 9, 19]
arr2 = [8, 11, 17, 21, 25]
print(merge_sorted_arr(arr1, arr2))
```

### №5 Слияние двух отсортированных массивов без дополнительных аллокаций

```python
def merge(arr1, arr2):
    p1 = len(arr1) - len(arr2) - 1
    p2 = len(arr2) - 1
    p3 = len(arr1) - 1
    while p2>=0:
        if p1>=0 and arr1[p1]>arr2[p2]:
            arr1[p3] = arr1[p1]
            p1-=1
        else:
            arr1[p3] = arr2[p2]
            p2-=1
        p3-=1
    return arr1
```

```python
arr1 = [3, 7, 9, 19]
arr2 = [8, 11, 17, 21, 25]
arr1.extend([0]*len(arr2))
print(merge(arr1, arr2))
```

### №6 Минимальный размер подмассива

```python
def minSubArr(nums, target):
    minLen = sys.maxsize
    left = 0
    curSum = 0
    for right in range(0, len(nums)):
        curSum += nums[right]
        while curSum>=target:
            windowSize = right - left + 1
            if windowSize < minLen:
                minLen = windowSize
            curSum -= nums[left]
            left+=1
    return minLen if minLen != sys.maxsize else 0
```

```python
nums = [3, 8, 9, 11, 16, 18, 19, 21]
print(minSubArr(nums, 100))
```

### №7 Сортировка массива из 0 и 1

```python
def sortBinaryArr(arr):
    left = 0
    right = len(arr)-1
    while left<right:
        if arr[left] == 1:
            arr[left], arr[right] = arr[right], arr[left]
            right-=1
        else:
            left+=1
    return arr
```

```python
nums = [0, 1, 1, 0, 0, 1, 1, 0, 1]
print(sortBinaryArr(nums))
```

### №8 Задача флага Нидерландов

```python
def sortColor(nums):
    low = 0
    mid = 0
    high = len(nums) - 1
    while mid<=high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low+=1
            mid+=1
        elif nums[mid] == 1:
            mid+=1
        elif nums[mid] == 2:
            nums[mid], nums[high] = nums[high], nums[mid]
            high-=1
    return nums
```

```python
nums = [2, 0, 2, 1, 1, 0]
print(sortColor(nums))
```

### №9 Передвинуть четные числа вперед

```python
def evenFirst(arr):
    evenIndex = 0
    for i in range(0, len(arr)):
        if arr[i] % 2 == 0:
            arr[i], arr[evenIndex] = arr[evenIndex], arr[i]
            evenIndex +=1
    return arr
```

```python
nums = [2, 4, 2, 1, 1, 5, 6, 7, 4, 4, 7]
print(evenFirst(nums))
```

### №10 Передвинуть нули в конец

```python
def zero(arr):
    Index = 0
    for i in range(0, len(arr)):
        if arr[i] != 0:
            arr[Index] = arr[i]
            Index +=1
    for i in range(Index, len(arr)):
        arr[i] = 0
    return arr
```

```python
nums = [2, 0, 2, 1, 0, 5, 6, 0, 4, 0, 7]
print(zero(nums))
```
